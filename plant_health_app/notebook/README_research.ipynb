## 1. Import Required Libraries

## 2. Tổng quan các thay đổi và lý do
- Chuẩn hóa cấu hình các optimizer: thêm momentum, weight decay, nesterov cho SGD; weight decay cho Adam/AdamW.
- Tách đường dẫn lưu mô hình, checkpoint, plots riêng cho từng optimizer để tránh ghi đè.
- Thêm MobileNetV2 và EfficientNet-B0 để so sánh với ResNet50.
- Điều chỉnh learning rate cho EfficientNet-B0 (nhạy cảm hơn).
- Chuẩn hóa batch_size, num_epochs, phát hiện GPU, bật AMP cho Google Colab.
- Thêm early stopping để tránh overfitting.
- Chuẩn hóa xử lý dữ liệu, mapping plant_type, loại bỏ dữ liệu không hợp lệ.
- Lưu lại group_classes.json để tiện tra cứu nhóm bệnh theo loại cây.
- Tối ưu hàm train_model: lưu checkpoint, best model, vẽ biểu đồ loss/accuracy.
- So sánh kết quả 3 models, vẽ biểu đồ tổng hợp và lưu vào plots.

## 3. Lý do chọn các tham số
- SGD với momentum và nesterov giúp mô hình hội tụ nhanh hơn, giảm hiện tượng mắc kẹt tại local minima.
- Weight decay giúp regularization, giảm overfitting.
- Adam/AdamW phù hợp với dữ liệu phức tạp, cần điều chỉnh weight decay hợp lý.
- Learning rate cao hơn Adam, nhưng cần giảm cho EfficientNet-B0.
